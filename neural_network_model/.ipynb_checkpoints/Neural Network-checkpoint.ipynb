```python
import os
import numpy as np
import tensorflow as tf
import pandas as pd
from tensorflow.keras import layers, models
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_squared_error
```

```python
#import the data from csv fike
csv_filename= "../datacleaning-pipeline/pnw_dataset.csv"
df = pd.read_csv(csv_filename)
```

```python
df['date']=pd.to_datetime(df['date'])
df = df.set_index('date')

df =df.sort_index()
```

```python
#seelction of the variables
input_columns = ['prcp(mm/day)', 'srad(W/m2)', 'tmax(C)', 'tmin(C)', 'vp(Pa)', 'lat', 'lon', 'drainage_area_km2', 'area_km2', 'elevation_m', 'slope_m_per_km','frac_forest', 'mean_precip_mm_day', 'mean_pet_mm_day', 'mean_temp_c']
target_column = ['runoff_mm_day']
```

```python
#Training (80%) vs Testing Period (20%) and Data
training_start, training_stop = '2009-10-01', '2013-09-30'
testing_start, testing_stop = '2013-10-01', '2014-09-30'

training_df = df.loc[training_start:training_stop]
testing_df = df.loc[testing_start:testing_stop]

X_training = training_df[input_columns].values
X_testing = testing_df[input_columns].values

y_training =np.log1p(training_df[target_column].values.reshape(-1,1))
y_testing =np.log1p(testing_df[target_column].values.reshape(-1,1))

print(f'Training array of size : {X_training.shape}')
```

```python
#Z-score scaling (Peter's advice)

scaler = StandardScaler()
X_training_scaled = scaler.fit_transform(X_training)
X_testing_scaled = scaler.transform(X_testing)
```

```python
#Modeling

model = models.Sequential([
    layers.Input(shape=(X_training_scaled.shape[1],)),
    layers.Dense(50, activation ='relu'),
    layers.Dense(50, activation='relu'),
    layers.Dense(25,activation='relu'),
    layers.Dense(1, activation='linear')])
```

```python
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()
```

```python
print('\n Beginning training... \n')
model.fit(X_training_scaled, y_training, epochs=100, batch_size=64, validation_split=0.1, verbose=1)
print('\n Training complete!\n')
```

```python
test_loss, test_mae = model.evaluate(X_testing_scaled, y_testing, verbose=0)
raw_mae= np.expm1(test_mae)
y_predict_log =model.predict(X_testing_scaled)
y_true_raw= np.expm1(y_predict_log)
y_predict_raw= np.expm1(y_testing)

y_true = y_true_raw.flatten()
y_predict = y_predict_raw.flatten()

print(f'Test MSE (Log Loss):{test_loss:.4f}')
print(f'Test MAE (Log MAE):{test_mae:.4f}')
print(f'Test MAE (Raw mm/day):{raw_mae:.4f}')
```

```python
nse_score = r2_score(np.expm1(y_testing), np.expm1(model.predict(X_testing_scaled)))
rmse_score =np.sqrt(mean_squared_error(y_true,y_predict))
print(f'This model has a NSE score of {nse_score:.4f}')
print(f'This model has a RMSE Score of {rmse_score:.4f}')
```
